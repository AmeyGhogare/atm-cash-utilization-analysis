## 2. Data Cleaning

```python
cleaned_df = df.copy()

print("Original dataset shape:", cleaned_df.shape)
```

```python
cleaned_df.columns = (
    cleaned_df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("Column names standardized successfully.")
```

```python
before_duplicates = cleaned_df.shape[0]

cleaned_df = cleaned_df.drop_duplicates()

after_duplicates = cleaned_df.shape[0]

print("Records before duplicate removal:", before_duplicates)
print("Records after duplicate removal:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
print("Total missing values:", cleaned_df.isnull().sum().sum())
```

```python
categorical_columns = [
    "Day_of_Week",
    "Time_of_Day",
    "Location_Type",
    "Weather_Condition"
]

for column in categorical_columns:
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    print(cleaned_df[column].value_counts())
```

```python
print("Holiday Flag:")
print(cleaned_df["Holiday_Flag"].value_counts())

print("\nSpecial Event Flag:")
print(cleaned_df["Special_Event_Flag"].value_counts())
```

```python
numerical_columns = [
    "Total_Withdrawals",
    "Total_Deposits",
    "Previous_Day_Cash_Level",
    "Nearby_Competitor_ATMs",
    "Cash_Demand_Next_Day"
]

for column in numerical_columns:
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    print("Minimum:", cleaned_df[column].min())
    print("Maximum:", cleaned_df[column].max())
```

```python
atm_counts = cleaned_df["ATM_ID"].value_counts()

print("Total unique ATMs:", cleaned_df["ATM_ID"].nunique())

print("\nRecords per ATM:")
print(atm_counts)
```

```python
print("Minimum date:", cleaned_df["Date"].min())
print("Maximum date:", cleaned_df["Date"].max())
```

```python
cleaned_df["Year"] = cleaned_df["Date"].dt.year
cleaned_df["Month"] = cleaned_df["Date"].dt.month
cleaned_df["Month_Name"] = cleaned_df["Date"].dt.month_name()
cleaned_df["Quarter"] = cleaned_df["Date"].dt.quarter
```

```python
print("========== CLEANED DATASET ==========")
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Unique ATMs:", cleaned_df["ATM_ID"].nunique())
```

### 3.10 Cash Utilization Metrics

```python
cleaned_df["Net_Cash_Flow"] = (
    cleaned_df["Total_Withdrawals"]
    - cleaned_df["Total_Deposits"]
)

cleaned_df["Cash_Utilization_Rate"] = (
    cleaned_df["Total_Withdrawals"]
    / cleaned_df["Previous_Day_Cash_Level"]
) * 100
```

```python
print(cleaned_df[
    [
        "ATM_ID",
        "Total_Withdrawals",
        "Total_Deposits",
        "Previous_Day_Cash_Level",
        "Net_Cash_Flow",
        "Cash_Utilization_Rate"
    ]
].head(10))
```

```python
cash_summary = {
    "Total Withdrawals": cleaned_df["Total_Withdrawals"].sum(),
    "Total Deposits": cleaned_df["Total_Deposits"].sum(),
    "Average Withdrawals": cleaned_df["Total_Withdrawals"].mean(),
    "Average Deposits": cleaned_df["Total_Deposits"].mean(),
    "Average Cash Level": cleaned_df["Previous_Day_Cash_Level"].mean(),
    "Average Next-Day Demand": cleaned_df["Cash_Demand_Next_Day"].mean(),
    "Average Utilization Rate": cleaned_df["Cash_Utilization_Rate"].mean()
}

for metric, value in cash_summary.items():
    print(f"{metric}: {value:,.2f}")
```

```python
atm_performance = (
    cleaned_df
    .groupby("ATM_ID")
    .agg(
        Total_Withdrawals=("Total_Withdrawals", "sum"),
        Total_Deposits=("Total_Deposits", "sum"),
        Average_Cash_Level=("Previous_Day_Cash_Level", "mean"),
        Average_Cash_Demand=("Cash_Demand_Next_Day", "mean"),
        Average_Utilization=("Cash_Utilization_Rate", "mean"),
        Records=("ATM_ID", "count")
    )
    .sort_values("Total_Withdrawals", ascending=False)
)

atm_performance.head(10)
```

```python
# Identify High-Demand ATM Records
demand_threshold = cleaned_df["Cash_Demand_Next_Day"].quantile(0.75)

high_demand_records = cleaned_df[
    cleaned_df["Cash_Demand_Next_Day"] >= demand_threshold
]

print("75th percentile demand threshold:", round(demand_threshold, 2))
print("High-demand records:", len(high_demand_records))
```

```python
high_demand_atms = (
    high_demand_records
    .groupby("ATM_ID")
    .agg(
        Average_Demand=("Cash_Demand_Next_Day", "mean"),
        Average_Cash_Level=("Previous_Day_Cash_Level", "mean"),
        Total_Withdrawals=("Total_Withdrawals", "sum"),
        Records=("ATM_ID", "count")
    )
    .sort_values("Average_Demand", ascending=False)
)

display(high_demand_atms.head(10))
```

```python
# ATM Risk Indicator
demand_threshold = cleaned_df["Cash_Demand_Next_Day"].quantile(0.75)
cash_level_threshold = cleaned_df["Previous_Day_Cash_Level"].quantile(0.25)
```

```python
cleaned_df["Cash_Risk"] = np.where(
    (cleaned_df["Cash_Demand_Next_Day"] >= demand_threshold) &
    (cleaned_df["Previous_Day_Cash_Level"] <= cash_level_threshold),
    "High Risk",
    "Normal"
)
```

```python
print(cleaned_df["Cash_Risk"].value_counts())
```

```python
# High-Risk ATM Summary
risk_summary = (
    cleaned_df[cleaned_df["Cash_Risk"] == "High Risk"]
    .groupby("ATM_ID")
    .agg(
        High_Risk_Records=("ATM_ID", "count"),
        Average_Demand=("Cash_Demand_Next_Day", "mean"),
        Average_Cash_Level=("Previous_Day_Cash_Level", "mean"),
        Total_Withdrawals=("Total_Withdrawals", "sum")
    )
    .sort_values("High_Risk_Records", ascending=False)
)

display(risk_summary.head(10))
```
